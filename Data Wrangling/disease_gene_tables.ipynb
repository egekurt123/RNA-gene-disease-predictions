```python
import pandas as pd
import numpy as np
```

# Mito predictions

```python
mito_predictions = pd.read_csv('../supplementary_tables/Supplementary_Table_S7_Mito_predictions.tsv', sep='\t')
mito_predictions
```

```python
print(mito_predictions[mito_predictions['emb'] == 'OMICS'].shape[0])
print(mito_predictions[mito_predictions['emb'] == 'STRING'].shape[0])
print(mito_predictions[mito_predictions['emb'] == 'STRING_EXP'].shape[0])
```

```python
mito_predictions_processed = mito_predictions.loc[mito_predictions['emb'] == 'STRING_EXP', ['gene_id', 'target']]
mito_predictions_processed.to_csv('../preprocessed_data/mito_genes.csv', index=False)
```
